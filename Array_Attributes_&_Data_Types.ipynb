# Array Attributes & Data Types
## NumPy Notes

These notes explain the most important NumPy array attributes and data types with commented examples.


### `.shape`
Returns the dimensions of an array.

- 1D → `(n,)`
- 2D → `(rows, columns)`
- 3D → `(depth, rows, columns)`

```
import numpy as np

# Create a 2D array
arr = np.array([[10, 20, 30],
                [40, 50, 60]])

# Returns (rows, columns)
print(arr.shape)
```

### `.ndim`
Returns the number of dimensions (axes).

```
import numpy as np

arr1 = np.array([1,2,3])      # 1D
arr2 = np.array([[1,2],[3,4]])# 2D

print(arr1.ndim)
print(arr2.ndim)
```

### `.size`
Returns the total number of elements.

```
import numpy as np

arr = np.array([[1,2,3],
                [4,5,6]])

# Total elements = rows × columns
print(arr.size)
```

### `.dtype`
Returns the data type of the array.

```
import numpy as np

arr = np.array([1,2,3])

# Display datatype
print(arr.dtype)
```

### `.itemsize`
Memory used by one element (in bytes).

```
import numpy as np

arr = np.array([1,2,3])

# Each int16 uses 2 bytes
print(arr.itemsize)
```

### `.nbytes`
Total memory used by the array.

Formula:

`nbytes = size × itemsize`


```
import numpy as np

arr = np.array([[1,2,3],
                [4,5,6]])

# Total memory consumed
print(arr.nbytes)
```

### `.T`
Interchanges rows and columns.

```
import numpy as np

arr = np.array([[1,2,3],
                [4,5,6]])

# Transpose the matrix
print(arr.T)
```

### `.flags`
Shows information about how the array is stored in memory.

```
import numpy as np

arr = np.array([1,2,3])

# Memory layout information
print(arr.flags)
```

### `.strides`
Shows how many bytes NumPy jumps in memory to move to the next element.

```
import numpy as np

arr = np.array([[1,2,3],
                [4,5,6]])

# Memory step size
print(arr.strides)
```

## Common NumPy Data Types

| Type | Description |
|------|-------------|
| `np.int8` | 8-bit Integer |
| `np.int16` | 16-bit Integer |
| `np.int32` | 32-bit Integer |
| `np.int64` | 64-bit Integer |
| `np.float16` | 16-bit Float |
| `np.float32` | 32-bit Float |
| `np.float64` | 64-bit Float |
| `np.bool_` | Boolean |
| `str` | Unicode String |


```
import numpy as np

int_arr = np.array([1,2,3], dtype=np.int16)
float_arr = np.array([1.5,2.5], dtype=np.float32)
bool_arr = np.array([True, False])
str_arr = np.array(["NumPy","Python"])

# Display datatypes
print(int_arr.dtype)
print(float_arr.dtype)
print(bool_arr.dtype)
print(str_arr.dtype)
```

## Automatic Type Promotion

NumPy automatically converts elements to a common compatible data type.

```
import numpy as np

# Integer becomes float
arr = np.array([1, 2, 3.5])
print(arr)
print(arr.dtype)
```

## `astype()`

Used to convert an existing array into another data type.

```
import numpy as np

arr = np.array([1,2,3])

# Convert integer array to float
float_arr = arr.astype(float)
print(float_arr)

# Convert float to integer (decimal part is truncated)
arr2 = np.array([1.9,2.5,3.8])
print(arr2.astype(int))

# Convert integers to boolean
arr3 = np.array([0,1,2])
print(arr3.astype(bool))

# Convert strings to integers
arr4 = np.array(["10","20","30"])
print(arr4.astype(int))
```
