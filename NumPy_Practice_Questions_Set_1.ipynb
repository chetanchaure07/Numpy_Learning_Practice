## Question 6

Find the shape, size, number of dimensions, and data type of a given array.


```python
# Write your code here
arr = np.arange(12).reshape(3, 4)
print(arr, '\n')
print("Shape:", arr.shape, '\n')
print("Size:", arr.size, '\n')
print("Data Type:", arr.dtype, '\n')
print("Dimention:", arr.ndim, '\n')
```

## Question 7

```python
# Write your code here
arr = np.arange(24)
print(arr, '\n')
ndim_arr = arr.reshape(4, 6)
print(ndim_arr)
```

## Question 8

## Question 9

Extract all elements greater than 50 from a random integer array.


```python
# Write your code here
arr = np.random.randint(30, 80, size = 16).reshape(4, 4)
print(arr, '\n')
print(arr[arr > 50])
```

## Question 10

Replace all negative values in an array with 0.


```python
# Write your code here
arr = np.arange(-5, 11)
print(arr, '\n')
arr[arr < 0] = 0
print(arr)
```

## Question 11

Create two 3×3 matrices and perform addition, subtraction, multiplication (element-wise), and matrix multiplication.


```python
# Write your code here
arr1 = np.random.randint(1, 50, size = 9).reshape(3, 3)
arr2 = np.random.randint(1, 50, size = 9).reshape(3, 3)
print(arr1 + arr2, '\n')
print(arr1 - arr2, '\n')
print(arr1 * 7, '\n')
print(arr1 * arr2, '\n')
print((arr1 / arr2).astype(int))
```

## Question 12

Find the maximum, minimum, mean, median, standard deviation, and sum of an array.


```python
# Write your code here
arr = np.arange(1, 17)
print(arr, '\n')
print("Maximum:", np.max(arr), '\n')
print("Minimun:", np.min(arr), '\n')
print("Mean:", np.mean(arr), '\n')
print("Standard deviation:", np.std(arr), '\n')
print("Sum of array:", np.sum(arr))
```
