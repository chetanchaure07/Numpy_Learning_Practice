# NumPy - Part 1: Introduction

## What is NumPy?

**NumPy (Numerical Python)** is a powerful Python library used for numerical computing. It provides a fast and efficient way to work with arrays and perform mathematical operations.

NumPy is one of the core libraries used in **Data Analysis, Data Science, Machine Learning, Artificial Intelligence, Scientific Computing, and Image Processing**.


## Why Do We Use NumPy?

We use NumPy because it offers several advantages over Python lists:

- **Faster Computation** – Operations are highly optimized and implemented in C.
- **Less Memory Usage** – NumPy arrays store data more efficiently than Python lists.
- **Mathematical Operations** – Perform element-wise arithmetic directly on arrays.
- **Multi-dimensional Arrays** – Easily work with 1D, 2D, and higher-dimensional data.
- **Foundation for Data Science** – Libraries like Pandas, Scikit-learn, TensorFlow, and PyTorch are built on or work closely with NumPy.


## Import NumPy

The standard way to import NumPy is:

```python
import numpy as np
```

The alias `np` is widely used in the Python community.
