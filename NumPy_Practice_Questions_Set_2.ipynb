## Question 1

Create a 6×6 array containing numbers from 1 to 36.


```python
# Write your code here
arr = np.arange(1, 37).reshape(6, 6)
print(arr)
```

## Question 2

Extract the first row, last row, first column, last column, and main diagonal.


```python
# Write your code here
print("First Row:", arr[0], '\n')
print("Last Row:", arr[-1], '\n')
print("First Column:", arr[ : , 0], '\n')
print("Last Column:", arr[ : , -1], '\n')
print("Diagonal:", np.diag(arr))
```

## Question 3

Extract all even numbers using Boolean indexing.


```python
# Write your code here
print(arr % 2 == 0)
```

## Question 4

Replace every element divisible by 3 with -1 without using loops.


```python
# Write your code here
arr[arr % 3 == 0] = -1
print(arr)
```

## Question 5

Create two random (4×4) arrays and compare them using >, <, ==, and !=.


```python
# Write your code here
arr1 = np.random.randint(1, 101, size = 16).reshape(4, 4)
arr2 = np.random.randint(1, 101, size = 16).reshape(4, 4)
print(arr1 < arr2, '\n')
print(arr1 > arr2, '\n')
print(arr1 == arr2, '\n')
print(arr1 != arr2)
```

## Question 6

Find the indices of all elements greater than the mean of an array.


```python
# Write your code here
arr1 = np.array([4, 5, 10, 48, 41, 11, 3, 7, 32, 20])
print("Mean:", arr1.mean(),  "Or",  np.mean(arr1))
indices = np.where(arr1 > arr1.mean())[0]
print(indices)
```

## Question 7

Use `np.where()` to replace values greater than 50 with 100, otherwise 0.


```python
# Write your code here
arr2 = np.random.randint(30, 70, size = 36).reshape(6, 6)
print(np.where(arr2 > 50, 100, 0))
print(np.where(arr2 > 50))
```

## Question 8

Generate 20 equally spaced values between 5 and 100 using `linspace()`.

## Question 9

Concatenate two 3×3 arrays vertically and horizontally.


```python
# Write your code here
arr1 = np.arange(9).reshape(3, 3)
arr2 = np.arange(10, 19).reshape(3, 3)
print(np.concatenate((arr1, arr2), axis = 0), '\n')
print(np.concatenate((arr1, arr2), axis = 1))
```

## Question 10

Split a 6×6 array into three equal parts row-wise and column-wise.



```python
arr = np.arange(36).reshape(6, 6)
print(arr, '\n')
print(np.vsplit(arr, 3), '\n')
print(np.hsplit(arr, 3))
```

## Question 11

Find all unique values and their frequencies.


```python
# Write your code here
arr = np.random.randint(1, 10, size = 20).reshape(5, 4)
unique, count = np.unique(arr, return_counts = True)
print("Unique value with frequencies")
print(unique)
print(count)
```

## Question 13

Calculate cumulative sum (`cumsum`) and cumulative product (`cumprod`).


```python
# Write your code here
arr = np.array([1, 2, 3, 4, 5])

print("Array:", arr)
print("Cumulative Sum:", np.cumsum(arr))
print("Cumulative Product:", np.cumprod(arr))
```

## Question 14

Find the transpose of a matrix and verify the new shape.


```python
# Write your code here
arr = np.array([[1, 2, 3],
                [4, 5, 6]])

print("Original Matrix:")
print(arr, '\n')

print("Transpose:")
print(arr.T)
```

## Question 15
